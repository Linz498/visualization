```python
# libraries
from matplotlib import rc
import pandas as pd
 
# Data
r = [0,1,2,3,4]
raw_data = {'Division_1': [20, 1.5, 7, 10, 5], 'Division_2': [5, 15, 5, 10, 15],'Division_3': [2, 15, 18, 5, 10]}
df = pd.DataFrame(raw_data)
 
# From raw value to percentage
totals = [i+j+k for i,j,k in zip(df['Division_1'], df['Division_2'], df['Division_3'])]
Division_1 = [i / j * 100 for i,j in zip(df['Division_1'], totals)]
Division_2 = [i / j * 100 for i,j in zip(df['Division_2'], totals)]
Division_3 = [i / j * 100 for i,j in zip(df['Division_3'], totals)]
 
# plot
barWidth = 0.85
names = ('Robbery','Auto_Theft','Assault','Shooting','Theft')
# Create green Bars
plt.bar(r, Division_1, color='#b5ffb9', edgecolor='white', width=barWidth)
# Create orange Bars
plt.bar(r, Division_2, bottom=Division_1, color='#f9bc86', edgecolor='white', width=barWidth)
# Create blue Bars
plt.bar(r, Division_3, bottom=[i+j for i,j in zip(Division_1, Division_2)], color='#a3acff', edgecolor='white', width=barWidth)
 
# Custom x axis
plt.xticks(r, names)
plt.xlabel("Crime Type")
 
# Show graphic
plt.show()
```
